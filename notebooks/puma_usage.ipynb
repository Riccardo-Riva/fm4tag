```python
import puma
import inspect
```

```python
def get_all_init_params(cls):
    params = {}
    for klass in cls.__mro__:
        if klass is object:
            continue
        try:
            sig = inspect.signature(klass.__init__)
        except (ValueError, TypeError):
            continue
        for name, p in sig.parameters.items():
            if name == 'self' or p.kind is p.VAR_KEYWORD:
                continue
            if name not in params:
                params[name] = p
    return params
```

```python
# usage
for name, p in get_all_init_params(puma.RocPlot).items():
    print(name, p.default, p.kind)
```

```python
roc_plot = puma.RocPlot(
    n_ratio_panels=0,
    ylabel='Background rejection',
    xlabel=f'b-jet efficiency',
    figsize=(7, 7),
    y_scale=1.3,
    atlas_first_tag='Simulation Internal',
    atlas_second_tag='cacca',
    dpi=300,
)
```

```python
# add a custom y=x curve
roc_plot.add_roc(
    puma.Roc(
        [0, 1],
        [0, 1],
        n_test=1,
        rej_class='c',
        signal_class='b',
        label='tagger',
    ),
    reference=True,
)
```

```python
roc_plot.draw()
```
